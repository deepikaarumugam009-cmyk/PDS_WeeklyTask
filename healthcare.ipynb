```python
# 29. Conditional filtering

# Patients with Cancer
cancer_patients = df[df['Medical Condition'] == 'Cancer']
print(cancer_patients)
df['Medical Condition'].value_counts()
```

```python
# Age > 60 and Emergency
senior_emergency = df[(df['Age'] > 60) & (df['Admission Type'] == 'Emergency')]
print(senior_emergency)

# Billing Amount > 40000
high_bill = df[df['Billing Amount'] > 40000]
print(high_bill)

# Female patients with Diabetes
female_diabetes = df[(df['Gender'] == 'Female') & (df['Medical Condition'] == 'Diabetes')]
print(female_diabetes)

# Filter using.between()
middle_age = df[df['Age'].between(30, 50)]
print(middle_age)
```

```python
df["Length of Stay"]=(
    pd.to_datetime(df["Discharge Date"])-
    pd.to_datetime(df["Date of Admission"])
).dt.days
df[["Date of Admission", "Discharge Date","Length of Stay"]].head()
```

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])

df["Length of stay"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days

df[["Age", "Billing Amount", "Room Number", "Length of stay"]].corr()
```

```python
df["Admission Year"]=pd.to_datetime(
    df["Date of Admission"]
).dt.year
df[["Date of Admission","Admission Year"]].head()
```

```python
df.groupby("Admission Year")["Billing Amount"].mean()
```

```python
df["Admission Month"]=pd.to_datetime(
    df["Date of Admission"]
).dt.year
df[["Date of Admission","Admission Month"]].head()
```

```python
monthly_admission=(
    df.groupby("Admission Month")["Name"]
    .count()
    .reset_index()
)
monthly_admission

```

```python
plt.figure(figsize=(9,5))
sns.lineplot(
    data=monthly_admission,
    x="Admission Month",
    y="Name",
    marker="o"
)
plt.title("Monthly Number of Admission")
plt.xlabel("Admission Month")
plt.ylabel("Number of admission")
plt.show()
```

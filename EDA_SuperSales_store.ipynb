```python
df["Order Date "]=pd.to_datetime(df["Order Date"])
df["Ship Date "]=pd.to_datetime(df["Ship Date"])
```

```python
df[["Order Date","Ship Date"]].dtypes
```

```python
text_column=["Category","Sub-Category","Segment"]
for column in text_column:
    df[column]=(
        df[column]
        .astype(str)
        .str.strip()
        .str.title()
    )    
```

```python
for column in text_column:
    print("\n",column)
    print(df[column].unique())
```

```python
print("Final Shape:",df.shape)
print("\nTotal missing Row:",df.isnull().sum().sum())
print("Total Duplicate Rows:",df.duplicated().sum())
print("\nData Types:")
print(df.dtypes)
```

```python
numerical_features=[
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]
df[numerical_features].describe()
```

```python
summary=pd.DataFrame({
    "Mean":df[numerical_features].mean(),
    "Median":df[numerical_features].median(),
    "Mimimum":df[numerical_features].min(),
    "maximum":df[numerical_features].max(),
    "Standard Deviation":df[numerical_features].std()
})
summary
    
```

```python
category_sales=(
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
category_sales
```

```python
category_profit=(
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)
category_profit
```

```python
sub_category_profit=(
    df.groupby("Sub-Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)
sub_category_profit
```

```python
region_profit=(
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
)
region_profit
```

```python
plt.figure(figsize=(4,5))
sns.barplot(
    data=df,
    x="Category",
    y="Sales",
    estimator="sum"
)
plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
plt.figure(figsize=(4,5))
sns.barplot(
    data=df,
    x="Category",
    y="Profit",
    estimator="sum"
)
plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()
```

```python
plt.figure(figsize=(4,4))
sns.boxplot(
    data=df,
    y="Sales"
)
plt.title("Sales Distribution")
plt.ylabel("Sales")
plt.show()
```

```python
plt.figure(figsize=(4,4))
sns.boxplot(
    data=df,
    y="Profit"
)
plt.title("Profit Distribution")
plt.ylabel("Profit")
plt.show()
```

```python
plt.figure(figsize=(4,4))
sns.boxplot(
    data=df,
    x="Category",
    y="Profit"
)
plt.title("Profit Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()
```

```python
plt.figure(figsize=(7,6))
sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit"
)
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.show()
```

```python
df["Profit Margin"] = df["Profit"]/df["Sales"]*100
df[["Sales","Profit","Profit Margin"]].head()
```

```python
numerical_df=df.select_dtypes(include=np.number)
correlation_matrix=numerical_df.corr()
correlation_matrix
```

```python
plt.figure(figsize=(7,6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)
plt.title("Correlation Heatmap of Numerical Attributes")
plt.show()
```
